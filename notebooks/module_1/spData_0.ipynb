# I. Пространственные данные

Как устроены пространственные данные и как читать, изучать и получать их в Python.

## Темы

1. [**Модели пространственных данных**](spData_1.ipynb)

   Векторная и растровая модели данных, классы Shapely и GeoPandas.

2. [**Чтение векторных данных**](spData_2.ipynb)

   Форматы Shapefile, GeoJSON, GeoPackage и CSV с координатами; сохранение данных.

3. [**Изучение набора данных**](spData_3.ipynb)

   Геометрия, система координат, атрибуты и охват набора данных.

4. [**Экспорт данных из OSM**](spData_4.ipynb)

   Выгрузка объектов OpenStreetMap по тегам с помощью `osmnx`.
