# Проект II: Картографирование демографических показателей

Проект выполняется **в командах по 2 человека**.

Выберите три соседних субъекта Российской Федерации и по данным Базы данных муниципальных образований (БДМО) Росстата проанализируйте, как изменились демографические показатели в их муниципалитетах.

## Данные

- **Границы муниципальных образований** — [геопортал НИУ ВШЭ](https://geoportal.hse.ru/portal/apps/sites/#/geodata/datasets/12e78b9c3bf04feea944f9d53eb3d079/about)
- **Показатели БДМО** от проекта «Если быть точным» — [полный набор данных и документация](https://tochno.st/datasets/bdmo), [данные по отдельным показателям](https://docs.google.com/spreadsheets/d/1Z66eXdgzyBg2NPV4cRzrTtv6BSZFVaAi/edit?pli=1&gid=1495188554#gid=1495188554)

## Что нужно сделать

1. **Выберите вместе три соседних субъекта** (у каждого должна быть общая граница хотя бы с одним из двух других) и **два года** для сравнения. Удобнее всего регионы, где от 20 до 50 муниципальных образований: на карте каждый из них будет хорошо виден.
2. **Проверьте данные** для выбранного региона и лет:
   - нет ли повторяющихся строк для одного муниципалитета и года;
   - для всех ли муниципалитетов есть данные за оба года;
   - не менялись ли за этот период границы муниципалитетов. Подсказка: если численность населения муниципалитета за несколько лет изменилась на 15–20 % и больше, скорее всего, к нему присоединили соседей.

   Проверку проводите для каждого из трёх субъектов. Если в каком-то регионе много преобразований, выберите другие годы или другой регион.

3. **Выберите два разных показателя** — сделайте необходимые вычисления

4. **Постройте карты** каждого показателя за оба года!
5. **Сформулируйте три ключевых вывода** по каждому показателю.
6. _(Опционально)_ **Проверьте, есть ли связь** между показателями или их динамикой. Если есть, то какая?
7. **Сделайте общий вывод:** что происходит в трёх регионах? Чем они похожи и чем различаются?

При построении карт обратите внимание на:

- единую легенду для карт, которые сравниваются между собой;
- выбор типа цветовой палитры;
- соответствие способа изображения типу данных;
- метод классификации: какой вы выбираете и почему?
- выбросы: проверьте муниципалитеты с аномальными значениями, прежде чем делать по ним выводы.

## Что должно быть на каждой карте

- **Заголовок:** показатель, территория и год (для карты изменений — период, например «2012–2020»).
- **Легенда** с единицами измерения (‰, %, чел./км²) и границами классов. У карт за два года — одна и та же шкала.
- **Границы субъектов** поверх муниципалитетов, чтобы было видно, где кончается один регион и начинается другой.
- **Муниципалитеты без данных** — отдельным нейтральным цветом и отдельной строкой в легенде, а не пустым местом.
- **Подпись с источником:** откуда данные (БДМО Росстата, «Если быть точным») и на какой год границы.

## Критерии оценивания (max. 10 баллов)

2. Реалистичность результата (2 балла)
3. Анализ и интерпретация (2 балла)
4. Читаемая и корректная визуализация (2 балла)
5. Структурированное оформление итоговой работы (2 балла)
