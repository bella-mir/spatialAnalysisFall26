# Проект I: Оценка доли застроенной территории и структура землепользования

Выберите район или город и по данным **OpenStreetMap** оцените:

- долю застроенной территории;
- структуру землепользования (площади и доли типов `landuse`).

Результаты покажите на картах и диаграмме, сделайте выводы о характере территории.

## Критерии оценивания (max. 10 баллов)

1. Корректность работы с пространственными данными, включая выбор и обоснование системы координат (2 балла)
2. Реалистичность результата (2 балла)
3. Анализ и интерпретация (2 балла)
4. Читаемая и корректная визуализация (2 балла)
5. Структурированное оформление итогового ноутбука (2 балла)
