# IV. Основные инструменты геообработки

Основные операции пространственного анализа векторных данных.


## Темы

1. [**Базовые операции с геометрией**](geoprocessing_1.ipynb)

   Буфер, слияние, обрезка и наложение.

2. [**Пространственные отношения**](geoprocessing_2.ipynb)

   Топологические предикаты и пространственная фильтрация.

3. [**Пространственное объединение и агрегирование**](geoprocessing_3.ipynb)

   `sjoin`, подсчёт объектов и статистика по атрибутам; проблема изменяемой единицы (MAUP).

4. [**Агрегирование точечных данных по регулярной сетке**](geoprocessing_4.ipynb)

   Плотность объектов в ячейках сетки и влияние размера ячейки.

5. [**Зоны ближайшего объекта и индекс ближайшего соседа**](geoprocessing_5.ipynb)

   Диаграмма Вороного и индекс ближайшего соседа (NNI).

6. [**Практикум: оценка численности населения**](geoprocessing_6.ipynb)

   Оценка численности жителей по данным о жилом фонде и карта плотности населения.
