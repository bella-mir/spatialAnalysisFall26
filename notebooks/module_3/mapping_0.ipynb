# III. Картографирование статистических данных

Как корректно показать статистические показатели на карте и сравнить их.


## Темы

1. [**Картограммы и методы классификации**](mapping_1.ipynb)

   Относительные показатели, цветовые палитры и методы классификации.

2. [**Подготовка данных БДМО**](mapping_2.ipynb)

   Муниципальная статистика Росстата: преобразование таблиц и объединение с границами по ОКТМО.

3. [**Сравнение карт и показателей**](mapping_3.ipynb)

   Единая легенда, карта изменений, корреляция показателей, сохранение карты.
